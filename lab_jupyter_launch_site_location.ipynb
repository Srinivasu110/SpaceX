```python
# TASK 3: Calculate the distances between a launch site to its proximities
# Import required libraries
from math import sin, cos, sqrt, atan2, radians
import folium
from folium.plugins import MousePosition
from folium.features import DivIcon

# Define the function to calculate the distance between two lat/lon points using Haversine formula
def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance

# Example launch site coordinates (CCAFS LC-40)
launch_site_lat = 28.562302
launch_site_lon = -80.577356

# Example coordinates of different proximity points (Coastline, Railway)
coastline_lat = 28.56367
coastline_lon = -80.57163
railway_lat = 28.5725
railway_lon = -80.5897

# Calculate the distances to the proximity points
distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
distance_railway = calculate_distance(launch_site_lat, launch_site_lon, railway_lat, railway_lon)

# Print out the calculated distances
print(f"Distance from launch site to coastline: {distance_coastline:.2f} km")
print(f"Distance from launch site to railway: {distance_railway:.2f} km")

# Create a Folium map centered around the launch site (e.g., CCAFS LC-40)
site_map = folium.Map(location=[launch_site_lat, launch_site_lon], zoom_start=12)

# Add a marker for the launch site
folium.Marker(
    [launch_site_lat, launch_site_lon],
    icon=DivIcon(icon_size=(20, 20), icon_anchor=(0, 0),
                html='<div style="font-size: 12px; color:#d35400;"><b>Launch Site</b></div>')
).add_to(site_map)

# Add a marker for the coastline with the distance label
folium.Marker(
    [coastline_lat, coastline_lon],
    icon=DivIcon(icon_size=(20, 20), icon_anchor=(0, 0),
                html='<div style="font-size: 12px; color:#d35400;"><b>{:.2f} KM to Coastline</b></div>'.format(distance_coastline))
).add_to(site_map)

# Add a marker for the railway with the distance label
folium.Marker(
    [railway_lat, railway_lon],
    icon=DivIcon(icon_size=(20, 20), icon_anchor=(0, 0),
                html='<div style="font-size: 12px; color:#d35400;"><b>{:.2f} KM to Railway</b></div>'.format(distance_railway))
).add_to(site_map)

# Add PolyLines (lines) between the launch site and each proximity point
folium.PolyLine(
    locations=[[launch_site_lat, launch_site_lon], [coastline_lat, coastline_lon]],
    weight=2, color='blue', opacity=0.6
).add_to(site_map)

folium.PolyLine(
    locations=[[launch_site_lat, launch_site_lon], [railway_lat, railway_lon]],
    weight=2, color='green', opacity=0.6
).add_to(site_map)

# Optional: Add MousePosition to the map for live coordinates
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright', separator=' Long: ', empty_string='NaN', lng_first=False,
    num_digits=20, prefix='Lat:', lat_formatter=formatter, lng_formatter=formatter
)
site_map.add_child(mouse_position)

# Display the map (this will work in Jupyter Notebooks or similar environments)
site_map
```

```python
# Create a marker with distance to a closest city, railway, highway, etc.
# Draw a line between the marker to the launch site
# Import required libraries
from math import sin, cos, sqrt, atan2, radians
import folium
from folium.plugins import MousePosition
from folium.features import DivIcon

# Define the function to calculate the distance between two lat/lon points using Haversine formula
def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance

# Example launch site coordinates (CCAFS LC-40)
launch_site_lat = 28.562302
launch_site_lon = -80.577356

# Create a Folium map centered around the launch site
site_map = folium.Map(location=[launch_site_lat, launch_site_lon], zoom_start=12)

# Add a marker for the launch site
folium.Marker(
    [launch_site_lat, launch_site_lon],
    icon=DivIcon(icon_size=(20, 20), icon_anchor=(0, 0),
                html='<div style="font-size: 12px; color:#d35400;"><b>Launch Site</b></div>')
).add_to(site_map)

# Example coordinates for city, railway, and highway
# In this case, you should manually explore the map using MousePosition and input coordinates
city_lat = 28.5736
city_lon = -80.648
railway_lat = 28.585
railway_lon = -80.600
highway_lat = 28.550
highway_lon = -80.550

# Calculate the distances from the launch site to each of the proximity points
distance_city = calculate_distance(launch_site_lat, launch_site_lon, city_lat, city_lon)
distance_railway = calculate_distance(launch_site_lat, launch_site_lon, railway_lat, railway_lon)
distance_highway = calculate_distance(launch_site_lat, launch_site_lon, highway_lat, highway_lon)

# Add markers for the city, railway, and highway with distance labels
folium.Marker(
    [city_lat, city_lon],
    icon=DivIcon(icon_size=(20, 20), icon_anchor=(0, 0),
                html='<div style="font-size: 12px; color:#d35400;"><b>{:.2f} KM to City</b></div>'.format(distance_city))
).add_to(site_map)

folium.Marker(
    [railway_lat, railway_lon],
    icon=DivIcon(icon_size=(20, 20), icon_anchor=(0, 0),
                html='<div style="font-size: 12px; color:#d35400;"><b>{:.2f} KM to Railway</b></div>'.format(distance_railway))
).add_to(site_map)

folium.Marker(
    [highway_lat, highway_lon],
    icon=DivIcon(icon_size=(20, 20), icon_anchor=(0, 0),
                html='<div style="font-size: 12px; color:#d35400;"><b>{:.2f} KM to Highway</b></div>'.format(distance_highway))
).add_to(site_map)

# Add PolyLines (lines) between the launch site and each proximity point (city, railway, highway)
folium.PolyLine(
    locations=[[launch_site_lat, launch_site_lon], [city_lat, city_lon]],
    weight=2, color='blue', opacity=0.6
).add_to(site_map)

folium.PolyLine(
    locations=[[launch_site_lat, launch_site_lon], [railway_lat, railway_lon]],
    weight=2, color='green', opacity=0.6
).add_to(site_map)

folium.PolyLine(
    locations=[[launch_site_lat, launch_site_lon], [highway_lat, highway_lon]],
    weight=2, color='red', opacity=0.6
).add_to(site_map)

# Optional: Add MousePosition to the map for live coordinates
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright', separator=' Long: ', empty_string='NaN', lng_first=False,
    num_digits=20, prefix='Lat:', lat_formatter=formatter, lng_formatter=formatter
)
site_map.add_child(mouse_position)

# Display the map (this will work in Jupyter Notebooks or similar environments)
site_map
```
